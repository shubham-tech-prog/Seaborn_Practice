# 2 . Categorical Data Plot 

```python
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = sns.load_dataset("tips")
df
```

```python
sns.countplot(x = df["sex"])  # x is for to make in on x axis 
```

```python
sns.countplot(x = df["sex"], hue = df["smoker"])  # hue is a parameter used to assign different colors to data points based on a categorical variable.
```

# barplot()

```python
sns.barplot(x = df["sex"], y = df["total_bill"])
```

```python
import numpy as np 
sns.barplot(x = df["sex"] , y = df["tip"], estimator= np.sum ) # It gives the total sums

# An estimator is a function that aggregates multiple data points into a single value, such as the mean or median.
```

```python
df
```

# *** Stripplot() and swarmplot() ***

```python
sns.stripplot(x = "day" , y = "total_bill" , data = df)
```
